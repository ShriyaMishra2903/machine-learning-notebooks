```python
# Variable Declaration

x = data[["Height"]]  # input variable  # Always a data frame # 2 dimensional 
y = data["Weight"]  # output variable
```

```python
y
```

```python
# Split the data into train and test
```

```python
# scikit learn: a library to build the ml model
!pip install scikit-learn
```

```python
from sklearn.model_selection import train_test_split   
x_train, x_test, y_train, y_test = train_test_split(x,y, test_size = 0.30, random_state = 5)
```

```python
x_train.shape
```

```python
x_test.shape
```

```python
y_train.shape
```

```python
y_test.shape
```

```python
# Calling the model _ regression model 
```

```python
from sklearn.linear_model import LinearRegression
model = LinearRegression()
```

```python
model
```

```python
# train the model
```

```python
model.fit(x_train, y_train)
```

```python
model.coef_
```

```python
model.intercept_
```

```python
y_pred = model.predict(x_test)
```

```python
from sklearn.metrics import r2_score
```

```python
print("Accracy is : ", np.round(r2_score(y_pred, y_test)*100,2) , "%")
```

```python
r2_score(y_pred, y_test)
```
